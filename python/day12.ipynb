```python
#amazon shopping cart using list
cart = []

def add_item(item):
    cart.append(item)
    print(f"{item} has been added to your cart.\n")

def remove_item(item):
    if item in cart:
        cart.remove(item)
        print(f"{item} has been removed\n")
    else:
        print(f"{item} not found in your cart.\n")

def total_items():
    return len(cart)


def display_cart():
    print(f"\nTotal items in cart: {total_items()}")
    for item in cart:
        print(item, end=" ")
    print("\n")

def main():
    while True:
        print("\n1. Add item to cart")
        print("2. Remove item from cart")
        print("3. View cart")
        print("4. Exit")
        choice = input("Enter your choice: ")

        if choice == '1':
            item = input("Enter the item to add: ")
            add_item(item)
        elif choice == '2':
            item = input("Enter the item to remove: ")
            remove_item(item)
        elif choice == '3':
            display_cart()
        elif choice == '4':
            print("Thank you for shopping with us!\n")
            break
        else:
            print("Invalid choice. Please try again.")

main()
```

```python
#create library management system using list

books = ["Book 1", "Book 2", "Book 3"]
backup_books = books.copy()

def add_book(book):
    books.append(book)
    print(f"{book} has been added to the library.\n")

def search_book(book):
    if book in books:
        print(f"{book} is available in the library.\n")
    else:
        print(f"{book} is not available in the library.\n")

def display_books():
    print(f"\nTotal books in library: {len(books)}")
    for book in books:
        print(book, end=" ")
    print("\n")

def issue_book(book):
    if book in books:
        books.remove(book)
        print(f"{book} has been issued.\n")
    else:
        print(f"{book} is not available for issue.\n")

def return_book(book):
    if book in backup_books:
        backup_books.remove(book)
        books.append(book)
        print(f"{book} has been returned to the library.\n")
    else:
        print(f"{book} is not from this library.\n")

def sort_books():
    books.sort()
    print("Books sorted in ascending order.\n")

def update_backup():
    global backup_books
    backup_books = books.copy()
    print("Backup updated.\n")

def main_library():
    while True:
        print("\n1. Add book to library")
        print("2. Search for a book")
        print("3. Display all books")
        print("4. Issue a book")
        print("5. Return a book")
        print("6. Sort books")
        print("7. Update backup")
        print("8. Exit")
        choice = input("Enter your choice: ")

        if choice == '1':
            book = input("Enter the book to add: ")
            add_book(book)
        elif choice == '2':
            book = input("Enter the book to search: ")
            search_book(book)
        elif choice == '3':
            display_books()
        elif choice == '4':
            book = input("Enter the book to issue: ")
            issue_book(book)
        elif choice == '5':
            book = input("Enter the book to return: ")
            return_book(book)
        elif choice == '6':
            sort_books()
        elif choice == '7':
            update_backup()
        elif choice == '8':
            print("Thank you for using the library management system!\n")
            break
        else:
            print("Invalid choice. Please try again.")

main_library()
```
