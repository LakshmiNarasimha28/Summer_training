```python
inputlist = []
for i in range(5):
    input_string = input("Enter a string: ")
    inputlist.append(input_string)

inputTuple = tuple(inputlist)
print("The tuple is:", inputTuple)

for i in range(len(inputTuple)):
    print(f"{i+1}, {inputTuple[i]}")
print()

for i in range(len(inputTuple)-1, -1, -1):
    print(f"{i+1}, {inputTuple[i]}")
```

```python
set1 = {2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8}

print(set1|set2)  # Union
print(set1&set2)  # Intersection
print(set1-set2)  # Difference
print(set1^set2)  # Symmetric Difference
```

```python
#remove duplicates from a list
list1 = [1, 2, 3, 4, 5, 1, 2, 3]
set1 = set(list1)
print(set1)
```

```python
#find common values

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print(set1.intersection(set2))  # Intersection
print(set1&set2)
```

```python
temperatures = [23.5, 25.0, 22.8, 24.1, 26.3]
average_temp = sum(temperatures) / len(temperatures)
print(f"The average temperature is: {average_temp:.2f}°C")
max_temp = max(temperatures)
min_temp = min(temperatures)
print(f"Maximum temperature: {max_temp:.2f}°C")
print(f"Minimum temperature: {min_temp:.2f}°C")
```

```python
programming_languages = set()

for i in range(5):
    language = input("Enter a programming language: ")
    programming_languages.add(language)

print("Programming languages in alphabetical order:")
for language in sorted(programming_languages):
    print(language)
```
