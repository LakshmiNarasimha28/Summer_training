```python
#frequency of every character in a sentence
def char_frequency(sentence):
    frequency = {}
    for char in sentence:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1
    return frequency

sentence = input("Enter a sentence: ")
char_frequency(sentence)
```

```python
# remove duplicate roll numbers from input and preserve original order
def remove_duplicates(roll_numbers):
    seen = set()
    unique_roll_numbers = []
    for roll_number in roll_numbers:
        if roll_number not in seen:
            seen.add(roll_number)
            unique_roll_numbers.append(roll_number)
    return unique_roll_numbers

roll_numbers = [input('Enter a roll number:') for _ in range(15)]

remove_duplicates(roll_numbers)
```

```python
# design a mini social network data model supporting friends, mutual friends and freind suggestions.
data = {
    'Alice': {'friends': ['Bob', 'Charlie']},
    'Bob': {'friends': ['Alice', 'Charlie', 'David']},
    'Charlie': {'friends': ['Alice', 'Bob', 'David']},
    'David': {'friends': ['Bob', 'Charlie']}
}

def get_mutual_friends(user1, user2):
    friends1 = set(data[user1]['friends'])
    friends2 = set(data[user2]['friends'])
    return list(friends1.intersection(friends2))

def get_friend_suggestions(user):
    user_friends = set(data[user]['friends'])
    suggestions = set()
    for friend in user_friends:
        suggestions.update(data[friend]['friends'])
    suggestions.difference_update(user_friends)
    suggestions.discard(user)
    return list(suggestions)

def display_friends(user):
    return data[user]['friends']

while True:
    print("\nMenu:")
    print("1. Get Mutual Friends")
    print("2. Get Friend Suggestions")
    print("3. Display friends")
    print("4. Exit App")
    choice = input("Enter your choice: ")

    match choice:
        case '1':
            user1 = input("Enter first user's name: ")
            user2 = input("Enter second user's name: ")
            mutual_friends = get_mutual_friends(user1, user2)
            print(f"Mutual friends between {user1} and {user2}: {mutual_friends}")
        case '2':
            user = input("Enter user's name: ")
            suggestions = get_friend_suggestions(user)
            print(f"Friend suggestions for {user}: {suggestions}")
        case '3':
            user = input("Enter user's name: ")
            friends = display_friends(user)
            print(f"{user}'s friends: {friends}")
        case '4':
            break
        case _:
            print("Invalid choice. Please try again.")
```

```python
#recomendation engine to suggest products purchased by similar customers but not by current customer
products = {
    'Customer1': {'ProductA', 'ProductB', 'ProductC'},
    'Customer2': {'ProductB', 'ProductC', 'ProductD'},
    'Customer3': {'ProductA', 'ProductD', 'ProductE'},
    'Customer4': {'ProductB', 'ProductE', 'ProductF'}
}

current_customer = input("Enter current customer name: ")
current_customer_products = products.get(current_customer, set())

similar_customers = {customer: prods for customer, prods in products.items() if customer != current_customer and prods.intersection(current_customer_products)}

print(f"products bought by similar customers to {current_customer}:")
unique_products = set()
for customer, prods in similar_customers.items():
    unique_products.update(prods.difference(current_customer_products))

print(unique_products)
```

```python
#airline reservation model to detect double booked seats

booked_seats = {
    "seat_1" : {'Customer1'},
    "seat_2" : {'Customer2'},
    "seat_3" : {'Customer3', 'Customer5'},
    "seat_4" : {'Customer4'},
    "seat_6" : {},
    "seat_7" : {'Customer8'},
    "seat_8" : {'Customer9', 'Customer12'},
    "seat_9" : {}
}

def detect_double_booked_seats(booked_seats):
    double_booked = {}
    for seat, customers in booked_seats.items():
        if len(customers) > 1:
            double_booked[seat] = customers
    return double_booked

detect_double_booked_seats(booked_seats)
```

```python
#an online coding contest leaderboard supporting rank updates after every submission

scoreboard = {
    'Alice': 1500,
    'Bob': 1200,
    'Charlie': 1800
}

def update_scoreboard(scoreboard, participant, new_score):
    scoreboard[participant] = new_score
    sorted_scoreboard = dict(sorted(scoreboard.items(), key=lambda item: item[1], reverse=True))
    return sorted_scoreboard

update_scoreboard(scoreboard, 'Alice', 1600)

for _ in range(3):
    participant = input("Enter participant name: ")
    new_score = int(input("Enter new score: "))
    scoreboard = update_scoreboard(scoreboard, participant, new_score)
    print("Updated Scoreboard:", scoreboard)
```

```python
# a plagiarism checker that compares two assignments based upon unique words and similarity percentage

def plagiarism_checker(assignment1, assignment2):
    set1words = set(assignment1.split(" "))
    set2words = set(assignment2.split(" "))
    unique_words = set1words | set2words
    common_words = set1words & set2words
    similarity_percentage = (len(common_words) / len(unique_words)) * 100 if unique_words else 0
    return similarity_percentage

Assignment1 = "ArithmeticError is a built-in exception in Python that is raised when an error occurs during an arithmetic operation. It is a subclass of the Exception class and is used to indicate that an arithmetic operation has failed due to some reason, such as division by zero or overflow."

Assignment2 = "ArithmeticError is a exception in Python that is raised when an error occurs during an arithmetic operation. Example: Division by zero or overflow. It is a subclass of the Exception class and is used to indicate that an arithmetic operation has failed due to some reason."

similarity_percentage = plagiarism_checker(Assignment1, Assignment2)
print(f"Similarity Percentage: {similarity_percentage:.2f}%")
```

```python
#calculator using try except
while True:
    try:
        num1 = float(input("Enter first number: "))
        operator = input("Enter operator (+, -, *, /): ")
        num2 = float(input("Enter second number: "))

        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            if num2 == 0:
                raise ZeroDivisionError("Cannot divide by zero.")
            result = num1 / num2
        else:
            raise ValueError("Invalid operator.")

        print(f"Result: {result}")
    except ValueError as e:
        print(f"Error: {e}")
    except ZeroDivisionError as e:
        print(f"Error: {e}")
```
