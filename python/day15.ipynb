```python
with open("example.txt", "w") as file:
    file.write("Hello, World!\nThis is a new line.\n")

while True:
    user_input = input("Enter a line to append to the file (or type 'exit' to quit): ")
    if user_input.lower() == 'exit':
        break
    with open("example.txt", "a") as file:
        file.write(user_input + "\n")

with open("example.txt", "r") as file:
    content = file.read()
    print("Updated file content:")
    print(content) 
```

```python
count_lines = 0
with open("example.txt", "r") as file:
    for line in file:
        count_lines += 1

print(f"The file 'example.txt' has {count_lines} lines.")
```

```python
count_words = 0
with open("example.txt", "r") as file:
    for line in file:
        words = line.split()
        count_words += len(words)

print(f"The file 'example.txt' has {count_words} words.")
```

```python
search_word = input("Enter a word to search in the file: ")
found = False
with open("example.txt", "r") as file:
    for line in file:
        if search_word in line:
            found = True
            break

if found:
    print(f"The word '{search_word}' was found in the file.")
else:
    print(f"The word '{search_word}' was not found in the file.")
```

```python
#copy from one file to another file

with open("example.txt", "r") as file:
    content = file.read()
    
with open("copy.txt", "w") as file:
    file.write(content)

with open("copy.txt", "r") as file:
    print("Content of copy.txt:")
    print(file.read())
```

```python
with open("example.txt", "r") as file:
    content = file.read()
    content = content.upper()

with open("example.txt", "w") as file:
    file.write(content)

with open("example.txt", "r") as file:
    print("Content of example.txt after converting to uppercase:")
    print(file.read())
```

```python
with open("students_names.txt", "w") as file:
    file.write("Alice\n")
    file.write("Bob\n")
    file.write("Charlie\n")
    file.write("David\n")
    file.write("Eve\n")

with open("students_names.txt", "r") as file:
    print("List of student names:")
    print(file.read())
```

```python
#read alternate lines from a file
with open("students_names.txt", "r") as file:
    lines = file.readlines()
    alternate_lines = lines[::2]  # Get every second line
    print("Alternate lines from the file:")
    for line in alternate_lines:
        print(line.strip())
```
