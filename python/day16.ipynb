```python
#types of methods in a class

class MyClass:
    # instance method
    def instance_method(self):
        return "This is an instance method."

    # class method
    @classmethod
    def class_method(cls):
        return "This is a class method."

    # static method
    @staticmethod
    def static_method():
        return "This is a static method."
    
my_object = MyClass()

type(my_object)
```

```python
#create a book class with attributes title, author, and price. add a method to display the book details.
class Book:
    title = ""
    author = ""
    price = 0.0
    
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display_details(self):
        return f"Title: {self.title}, Author: {self.author}, Price: ${self.price}"  
    
book1 = Book("The Great Gatsby", "F. Scott Fitzgerald", 12.99)
book1.display_details()
```

```python
class Student:
    roll_no = ""
    name = ""
    marks = 0

    def __init__(self, roll_no, name, marks):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks

    def display_details(self):
        return f"Roll No: {self.roll_no}, Name: {self.name}, Marks: {self.marks}"
    
    def get_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        else:
            return "C"
        
student1 = Student("101", "Alice", 85)
student1.display_details()
student1.get_grade()
```

```python
#create rectanle class wit h attributes length and width. add a method to calculate area and perimeter of the rectangle.
class Rectangle:
    length = 0
    width = 0

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)
    

rectangle1 = Rectangle(5, 3)
rectangle1.area()
```

```python
#create an employee class with attributes name, id, and salary. add a method give_raise() that updates self.salary in place. 
class Employee:
    name = ""
    id = ""
    salary = 0.0

    def __init__(self, name, id, salary):
        self.name = name
        self.id = id
        self.salary = salary

    def give_raise(self, amount):
        self.salary += amount
        print(f"New salary for {self.name} (ID: {self.id}) is: ${self.salary}")

employee1 = Employee("John Doe", "E123", 50000)
employee1.give_raise(5000)
```

```python
#create a library class that holds a list of book objects from book class. add methods to add a book, and display all books in the library.
class Book:
    title = ""
    author = ""
    price = 0.0
    
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display_details(self):
        return f"Title: {self.title}, Author: {self.author}, Price: ${self.price}"  


class Library:
    books = []
    
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)
        print(f"Book '{book.title}' added to the library.")

    def display_books(self):
        if not self.books:
            print("No books in the library.")
        else:
            for book in self.books:
                print(book.display_details())

Book1 = Book("1984", "George Orwell", 9.99)
Book2 = Book("To Kill a Mockingbird", "Harper Lee", 7.99)
library = Library()
library.add_book(Book1)
library.add_book(Book2)
library.display_books()
```

```python
#encapsulation in python
class employee:
    name = ""
    _salary = 0.0  # private attribute

    def __init__(self, name, salary):
        self.name = name
        self._salary = salary

    def get_salary(self):
        return self._salary
    
    def set_salary(self, salary):
        if salary >= 0:
            self._salary = salary
        else:
            print("Salary cannot be negative.") 

    def apply_raise(self, percentage):
        if percentage > 0:
            salary = self._salary * (percentage / 100)
            self.set_salary(self._salary + salary)
        else:
            print("Raise percentage must be positive.")
```
