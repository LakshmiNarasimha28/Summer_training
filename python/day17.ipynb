```python
#create an abstract class vec wit an abstract metod fuel_type() create petrolCar and ElectricCar subclasses eac implementin fuel_type() differently

class Vechile:
    def fuel_type(self):
        raise NotImplementedError("Subclasses must implement this method")


class PetrolCar(Vechile):
    def fuel_type(self):
        return "Petrol"
    
class ElectricCar(Vechile):
    def fuel_type(self):
        return "Electric"


pcar = PetrolCar()
print(pcar.fuel_type())  # Output: Petrol

ecar = ElectricCar()
print(ecar.fuel_type())  # Output: Electric
```

```python
#create a base class Person with name and age. create Teacher and student subclasses each adding one additional attribute and overriding a common method Role() to describe who they are.

class Person:
    name = ""
    age = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def role(self):
        raise NotImplementedError("Subclasses must implement this method")
    
class Teacher(Person):
    subject = ""

    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def role(self):
        return f"Teacher of {self.subject}"
    
class Student(Person):
    roll_no = ""

    def __init__(self, name, age, roll_no):
        super().__init__(name, age)
        self.roll_no = roll_no

    def role(self):
        return f"Student with roll number {self.roll_no}"
    

Teacher1 = Teacher("Alice", 30, "Mathematics")
print(Teacher1.role())  # Output: Teacher of Mathematics

Student1 = Student("Bob", 20, "12345")
print(Student1.role())  # Output: Student with roll number 12345
```

```python
#create square and triangle classes with a method draw() that prints a different simple pattern example: using asterisks (*).put objects of both in a list and loop through, calling the draw() method on each object.

class Square:
    def draw(self):
        print("*****")
        print("*   *")
        print("*   *")
        print("*   *")
        print("*****")

class Triangle:
    def draw(self):
        print("  *  ")
        print(" * * ")
        print("*****")

square = Square()
triangle = Triangle()

for shape in [square, triangle]:
    shape.draw()
```

```python
# design a system to issue and return books. store book availability safely allow members to borrow and return books and calculate late fees if applicable.

class Book:
    def __init__(self, title):
        self.title = title
        self.is_available = True

    def borrow(self):
        if self.is_available:
            self.is_available = False
            return True
        else:
            return False

    def return_book(self):
        self.is_available = True

class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed_books = {}

    def borrow_book(self, book, days_borrowed):
        if book.borrow():
            self.borrowed_books[book.title] = days_borrowed
            print(f"{self.name} borrowed '{book.title}' for {days_borrowed} days.")
        else:
            print(f"'{book.title}' is not available for borrowing.")

    def return_book(self, book):
        if book.title in self.borrowed_books:
            days_borrowed = self.borrowed_books.pop(book.title)
            book.return_book()
            late_fee = self.calculate_late_fee(days_borrowed)
            print(f"{self.name} returned '{book.title}'. Late fee: ${late_fee:.2f}")
        else:
            print(f"{self.name} did not borrow '{book.title}'.")

    def calculate_late_fee(self, days_borrowed):
        allowed_days = 14  # Assume 14 days is the borrowing period
        late_days = max(0, days_borrowed - allowed_days)
        return late_days * 0.50  # Assume $0.50 per late day
    

book1 = Book("The Great Gatsby")
member1 = Member("John Doe")
member1.borrow_book(book1, 10)  # Borrowing within allowed days
member1.return_book(book1)  # Returning the book
```

```python
# create economy business and first class tickets calculate ticket fare on class and baggage weight print boarding details

class Ticket:
    def __init__(self, ticket_class, baggage_weight):
        self.ticket_class = ticket_class
        self.baggage_weight = baggage_weight

    def calculate_fare(self):
        base_fare = {
            "economy": 100,
            "business": 200,
            "first": 300
        }
        fare = base_fare.get(self.ticket_class.lower(), 0)
        baggage_fee = max(0, self.baggage_weight - 20) * 10  # $10 for each kg over 20kg
        return fare + baggage_fee

    def boarding_details(self):
        fare = self.calculate_fare()
        return f"Ticket Class: {self.ticket_class}, Baggage Weight: {self.baggage_weight}kg, Total Fare: ${fare:.2f}"
    

ticket1 = Ticket("Economy", 25)
print(ticket1.boarding_details())  
```

```python
#CUSTOMERS ORDER FOOD ITEMS GENERATE THE BILL APPLY GST AND DISCOUNTS AND SUPPORT DINEIN AND TAKEAWAY ORDERS DIFFERENTLY

class Customer:
    def __init__(self, name):
        self.name = name
        self.order_items = []
        self.is_dine_in = True

    def add_item(self, item, price):
        self.order_items.append((item, price))

    def set_order_type(self, dine_in):
        self.is_dine_in = dine_in

    def calculate_bill(self):
        subtotal = sum(price for _, price in self.order_items)
        gst = subtotal * 0.05  # 5% GST
        discount = subtotal * 0.10 if self.is_dine_in else 0  # 10% discount for dine-in
        total = subtotal + gst - discount
        return total

    def generate_bill(self):
        print(f"Customer: {self.name}")
        print("Order Items:")
        for item, price in self.order_items:
            print(f"{item}: ${price:.2f}")
        total = self.calculate_bill()
        print(f"Total Bill (including GST and discounts): ${total:.2f}")
    
customer1 = Customer("Alice")
customer1.add_item("Burger", 8.99)
customer1.generate_bill()

customer2 = Customer("Bob")
customer2.set_order_type(False)  # Takeaway order
customer2.add_item("Pizza", 12.50)
customer2.generate_bill()
```

```python
# store student details privately conduct an exam calculate marks, assign grades, and generate a report card for each student.
class Student:
    def __init__(self, name, student_id):
        self.__name = name
        self.__student_id = student_id
        self.__marks = []

    def add_mark(self, mark):
        self.__marks.append(mark)

    def calculate_average(self):
        if not self.__marks:
            return 0
        return sum(self.__marks) / len(self.__marks)

    def assign_grade(self):
        average = self.calculate_average()
        if average >= 80:
            return "A"
        elif average >= 60:
            return "B"
        else:
            return "C"

    def generate_report_card(self):
        print(f"Student Name: {self.__name}")
        print(f"Student ID: {self.__student_id}")
        print(f"Average Marks: {self.calculate_average():.2f}")
        print(f"Grade: {self.assign_grade()}")


student1 = Student("Alice", "S001")
student1.add_mark(85)
student1.add_mark(78)
student1.generate_report_card()
```
