```python
import numpy as np
```

```python
#student marks analysis system (usin numpy)
#dataset creation
rng = np.random.default_rng()

# 100 students, 5 subjects
student_marks = rng.integers(35, 101, size=(35, 5))

subjects = ["Math", "Science", "English", "Computer", "History"]
student_names = ["Aarav","Vivaan","Aditya","Arjun","Krishna","Rohan","Rahul","Karan","Yash","Harsh","Ananya","Diya","Priya","Aditi",
"Sneha","Kavya","Meera","Ishita","Riya","Neha","Akash","Siddharth","Manav","Nikhil","Varun","Tanvi","Sanya","Pooja","Muskan","Simran",
"Dev","Aryan","Shreya","Aniket","Palak"]

print(student_marks)
print(student_marks.shape)
```

```python
avg_marks = np.mean(student_marks, axis=1)
print(avg_marks)
```

```python
highest = np.max(student_marks)
lowest = np.min(student_marks)

print(highest)
print(lowest)
```

```python
std_dev = np.std(student_marks, 1)

print(std_dev)
```

```python
ranking = np.argsort(avg_marks)[::-1]

for i in range(len(student_names)):
    print(f"{student_names[i]}: Rank {ranking[i]+1}")
```

```python
class_average = np.mean(avg_marks)

top_students = np.where(avg_marks > class_average)[0]

print(*(student_names[i] for i in top_students), sep=", ")
```

```python
subject_average = np.mean(student_marks, axis=0)
subject_highest = np.max(student_marks, axis=0)
subject_lowest = np.min(student_marks, axis=0)

print(subject_average)
print(subject_highest)
print(subject_lowest)
```

```python
total_marks = np.sum(student_marks, axis=1)

percentage = (total_marks / 500) * 100

print(percentage)
```
